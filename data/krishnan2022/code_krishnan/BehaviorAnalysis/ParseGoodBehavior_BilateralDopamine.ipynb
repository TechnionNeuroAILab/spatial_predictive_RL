```python
fs, ax = plt.subplots(2, 2, sharex='all',sharey='all', dpi=100)
ax = ax.flatten()
with sns.axes_style('white'):
    for n, i in enumerate(['TaskSal1', 'TaskSal2', 'Task5CNO1', 'Task5CNO2']):
        ax[n].plot(A.running_data[i])
        ax[n].set_xlim((0, 12000))
        ax[n].set_title(A.TaskDict[i])
        sns.despine(ax=ax[n])
        ax[n].set_yticks((0, 0.6))
        ax[n].set_yticklabels((0, 200))
ax[0].set_ylabel('Track Length (cm)')
ax[-1].set_xlabel('Frames')
# fs.savefig('Sample Behavior.pdf', bbox_inches='tight')
```

```python
fs, ax = plt.subplots(1, dpi=100)
x = dict((l, A.actual_lap_time[l]) for l in ['TaskSal1', 'TaskSal2', 'Task5CNO1', 'Task5CNO2'])
with sns.axes_style('white'):
    labels, data = [*zip(*x.items())]
    labels = [A.TaskDict[i] for i in labels]
    bp = ax.boxplot(data, patch_artist=True)
    plt.xticks(range(1, len(labels) + 1), labels)
    for median in bp['medians']:
        median.set(color='black', linewidth=1)
sns.despine(ax=ax)
ax.set_ylabel('Time taken to complete a lap (seconds)')
ax.locator_params(axis='y', nbins=4)
fs.savefig('Sample Velocity.pdf', bbox_inches='tight')
```
